```python
def collapse_fasta(in_path, out_path):
    with open(in_path) as fi, open(out_path, "w") as fo:
        header = None
        seq_parts = []

        for line in fi:
            line = line.rstrip()
            if not line:
                continue
            if line.startswith(">"):
                # flush previous record
                if header is not None:
                    fo.write(f"{header}\n{''.join(seq_parts)}\n")
                header = line
                seq_parts = []
            else:
                # accumulate sequence fragments
                seq_parts.append(line)

        # write last record
        if header is not None:
            fo.write(f"{header}\n{''.join(seq_parts)}\n")
```

```python
if __name__ == "__main__":
    infile  = "multiline_input.fasta"
    outfile = "singleline_output.fasta"
    collapse_fasta(infile, outfile)
    print(f"Wrote single-line FASTA to {outfile}")
```
