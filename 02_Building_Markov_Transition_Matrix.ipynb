```python
import numpy as np
import pandas as pd
```

```python
def build_transition_matrix(sequence):
    # Define possible nucleotides
    nucleotides = ['A', 'C', 'G', 'T']

    # Initialize a count matrix
    count_matrix = {n1: {n2: 0 for n2 in nucleotides} for n1 in nucleotides}

    # Count transitions
    for i in range(len(sequence) - 1):
        current = sequence[i]
        next_ = sequence[i + 1]
        if current in nucleotides and next_ in nucleotides:
            count_matrix[current][next_] += 1

    # Convert counts to probabilities
    transition_matrix = {}
    for current in nucleotides:
        total = sum(count_matrix[current].values())
        transition_matrix[current] = {
            next_: count_matrix[current][next_] / total if total > 0 else 0.0
            for next_ in nucleotides
        }

    # Create and print the matrix as a DataFrame
    df = pd.DataFrame(transition_matrix).T  # I have taken transpose for better display
    print("Markov Transition Matrix:")
    print(df.round(7))
    return df
```
